# Imports



```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.optimize import fsolve

from matplotlib.animation import FuncAnimation

from IPython.display import HTML

import ipywidgets as widgets

from ipywidgets import interact

plt.rcParams['figure.dpi'] = 120
```

# Parámetros del mecanismo sintetizado



```python
# ═══════════════════════════════════════════════════════════════════════════════
# PARÁMETROS ÓPTIMOS
# ═══════════════════════════════════════════════════════════════════════════════

L1 = 3
L2 = 1.0000
# L3 = 3.1362
L3 = 3.2

L4 = 1.7

# Parámetros modificables

d_default = 3
gamma_default = np.deg2rad(-10)
```


---

# Clase FourBar



```python
# ═══════════════════════════════════════════════════════════════════════════════
# MODELO CINEMÁTICO
# ═══════════════════════════════════════════════════════════════════════════════

class FourBar:

    def __init__(self, L1, L2, L3, L4, d, gamma):

        self.L1 = L1
        self.L2 = L2
        self.L3 = L3
        self.L4 = L4

        self.d = d
        self.gamma = gamma

        self._z0 = np.array([0.3, 2.5])

    # ──────────────────────────────────────────────────────────────────────
    # Grashof
    # ──────────────────────────────────────────────────────────────────────

    def grashof(self):

        links = np.array([
            self.L1,
            self.L2,
            self.L3,
            self.L4
        ])

        s = np.min(links)
        l = np.max(links)

        others = np.sort(links)[1:3]

        cond = (s + l) <= np.sum(others)

        shortest = np.argmin(links)

        if cond:

            tipos = {
                0: 'double-crank',
                1: 'crank-rocker',
                2: 'double-rocker',
                3: 'rocker-crank'
            }

            tipo = tipos[shortest]

        else:

            tipo = 'triple-rocker'

        return {
            'grashof': cond,
            'type': tipo
        }

    # ──────────────────────────────────────────────────────────────────────
    # Ecuaciones de cierre
    # ──────────────────────────────────────────────────────────────────────

    def residuals(self, x, theta2):

        theta3, theta4 = x

        f1 = (
            self.L2*np.cos(theta2)
            + self.L3*np.cos(theta3)
            - self.L1
            - self.L4*np.cos(theta4)
        )

        f2 = (
            self.L2*np.sin(theta2)
            + self.L3*np.sin(theta3)
            - self.L4*np.sin(theta4)
        )

        return [f1, f2]

    # ──────────────────────────────────────────────────────────────────────
    # FK
    # ──────────────────────────────────────────────────────────────────────

    def FK(self, theta2):

        sol, info, ier, _ = fsolve(
            self.residuals,
            self._z0,
            args=(theta2,),
            xtol=1e-10,
            full_output=True
        )

        success = (
            ier == 1
            and np.max(np.abs(info['fvec'])) < 1e-8
        )

        if success:

            d3 = np.abs(sol[0] - self._z0[0])
            d4 = np.abs(sol[1] - self._z0[1])

            if d3 > np.pi/2 or d4 > np.pi/2:
                return sol, False

            self._z0 = sol.copy()

        return sol, success

    # ──────────────────────────────────────────────────────────────────────
    # Posiciones
    # ──────────────────────────────────────────────────────────────────────

    def joint_positions(self, theta2):

        (theta3, theta4), ok = self.FK(theta2)

        if not ok:
            return {'ok': False}

        O2 = np.array([0,0])

        O4 = np.array([self.L1, 0])

        A = O2 + self.L2*np.array([
            np.cos(theta2),
            np.sin(theta2)
        ])

        B = A + self.L3*np.array([
            np.cos(theta3),
            np.sin(theta3)
        ])

        P = A + self.d*np.array([
            np.cos(theta3 + self.gamma),
            np.sin(theta3 + self.gamma)
        ])

        return {
            'ok': True,
            'O2': O2,
            'A': A,
            'B': B,
            'O4': O4,
            'P': P,
            'theta3': theta3,
            'theta4': theta4
        }
```

# Visualizador interactivo de configuración



```python
# ═══════════════════════════════════════════════════════════════════════════════
# VISUALIZADOR ESTÁTICO
# ═══════════════════════════════════════════════════════════════════════════════

def visualizar_configuracion(theta_deg=45, d=2.99, gamma_deg=-9):

    gamma = np.deg2rad(gamma_deg)

    mec = FourBar(
        L1, L2, L3, L4,
        d=d,
        gamma=gamma
    )

    pts = mec.joint_positions(np.deg2rad(theta_deg))

    if not pts['ok']:
        print('Configuración inválida')
        return

    O2 = pts['O2']
    A = pts['A']
    B = pts['B']
    O4 = pts['O4']
    P = pts['P']

    fig, ax = plt.subplots(figsize=(8,8))

    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)

    # Ground
    ax.plot(
        [O2[0], O4[0]],
        [O2[1], O4[1]],
        'k--',
        lw=4,
        label='Bancada'
    )

    # Crank
    ax.plot(
        [O2[0], A[0]],
        [O2[1], A[1]],
        lw=5,
        color='tab:red',
        label='L2 — Manivela'
    )

    # Coupler
    ax.plot(
        [A[0], B[0]],
        [A[1], B[1]],
        lw=5,
        color='tab:blue',
        label='L3 — Acoplador'
    )

    # Rocker
    ax.plot(
        [O4[0], B[0]],
        [O4[1], B[1]],
        lw=5,
        color='tab:green',
        label='L4 — Balancín'
    )

    # Segmento AP
    ax.plot(
        [A[0], P[0]],
        [A[1], P[1]],
        ':',
        lw=3,
        color='orange',
        label='Segmento AP'
    )

    # Joints
    ax.plot(O2[0], O2[1], 'ks', ms=10)
    ax.plot(O4[0], O4[1], 'ks', ms=10)

    ax.plot(A[0], A[1], 'ko', ms=8)
    ax.plot(B[0], B[1], 'ko', ms=8)

    ax.plot(P[0], P[1], '*', ms=18, color='orange')

    # Texto
    texto = (
        f'L1 = {L1:.3f}\n'
        f'L2 = {L2:.3f}\n'
        f'L3 = {L3:.3f}\n'
        f'L4 = {L4:.3f}\n\n'
        f'd = {d:.3f}\n'
        f'γ = {gamma_deg:.2f}°'
    )

    ax.text(
        0.02,
        0.98,
        texto,
        transform=ax.transAxes,
        va='top',
        family='monospace',
        bbox=dict(facecolor='white', alpha=0.9)
    )

    ax.legend()

    ax.set_title('Configuración del mecanismo')

    plt.show()
```

# Sliders interactivos



```python
# ═══════════════════════════════════════════════════════════════════════════════
# SLIDERS INTERACTIVOS
# ═══════════════════════════════════════════════════════════════════════════════

interact(
    visualizar_configuracion,

    theta_deg=widgets.FloatSlider(
        min=0,
        max=360,
        step=1,
        value=45,
        description='θ₂'
    ),

    d=widgets.FloatSlider(
        min=0.5,
        max=5.0,
        step=0.01,
        value=d_default,
        description='d'
    ),

    gamma_deg=widgets.FloatSlider(
        min=-180,
        max=180,
        step=1,
        value=np.rad2deg(gamma_default),
        description='γ'
    )
)
```

# Animación completa del mecanismo



```python
# ═══════════════════════════════════════════════════════════════════════════════
# ANIMACIÓN COMPLETA
# ═══════════════════════════════════════════════════════════════════════════════

def animar_mecanismo(d=d_default, gamma_deg=-9.07):

    gamma = np.deg2rad(gamma_deg)

    mec = FourBar(
        L1, L2, L3, L4,
        d=d,
        gamma=gamma
    )

    # trayectoria completa
    curve = []

    for th in np.linspace(0, 2*np.pi, 360):

        pts = mec.joint_positions(th)

        if pts['ok']:
            curve.append(pts['P'])

    curve = np.array(curve)

    fig, ax = plt.subplots(figsize=(8,8))

    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)

    # xmin = curve[:,0].min()
    # xmax = curve[:,0].max()

    # ymin = curve[:,1].min()
    # ymax = curve[:,1].max()
    
    xmin = 0
    xmax = 6

    ymin = 0
    ymax = 2

    pad = 0

    ax.set_xlim(xmin-pad, xmax+pad)
    ax.set_ylim(ymin-pad, ymax+pad)

    ax.plot(
        curve[:,0],
        curve[:,1],
        '--',
        color='lightgray'
    )

    crank, = ax.plot([], [], lw=5, color='tab:red')
    coupler, = ax.plot([], [], lw=5, color='tab:blue')
    rocker, = ax.plot([], [], lw=5, color='tab:green')
    ground, = ax.plot([], [], 'k--', lw=3)

    AP_line, = ax.plot([], [], ':', lw=3, color='orange')

    trail, = ax.plot([], [], color='crimson', lw=2)

    pointP, = ax.plot([], [], '*', ms=16, color='orange')

    hist_x = []
    hist_y = []

    theta_vals = np.linspace(0, 2*np.pi, 360)

    def update(i):

        theta2 = theta_vals[i]

        pts = mec.joint_positions(theta2)

        if not pts['ok']:
            return ()

        O2 = pts['O2']
        A = pts['A']
        B = pts['B']
        O4 = pts['O4']
        P = pts['P']

        crank.set_data(
            [O2[0], A[0]],
            [O2[1], A[1]]
        )

        coupler.set_data(
            [A[0], B[0]],
            [A[1], B[1]]
        )

        rocker.set_data(
            [O4[0], B[0]],
            [O4[1], B[1]]
        )

        ground.set_data(
            [O2[0], O4[0]],
            [O2[1], O4[1]]
        )

        AP_line.set_data(
            [A[0], P[0]],
            [A[1], P[1]]
        )

        pointP.set_data([P[0]], [P[1]])

        hist_x.append(P[0])
        hist_y.append(P[1])

        trail.set_data(hist_x, hist_y)

        return (
            crank,
            coupler,
            rocker,
            ground,
            AP_line,
            pointP,
            trail
        )

    anim = FuncAnimation(
        fig,
        update,
        frames=len(theta_vals),
        interval=25,
        blit=True
    )

    plt.close()

    return HTML(anim.to_jshtml())
```


---

# Ejecutar animación



```python
animar_mecanismo(
    d=d_default,
    gamma_deg=np.rad2deg(gamma_default)
)
```

# Exploración CAD recomendada

## Recomendaciones

Durante la exploración interactiva:

### Variar principalmente:

- `d`
- `γ`

porque son los parámetros que más alteran:

- amplitud,
- lift,
- curvatura,
- y comportamiento biomecánico.

---

## Qué observar

### 1. Continuidad

Verificar:
- ausencia de flips,
- ausencia de branch switching,
- continuidad completa.

---

### 2. Lift

Identificar:
- dónde ocurre máxima elevación,
- amplitud del stroke.

---

### 3. Manufacturabilidad

Evitar:
- segmentos excesivamente largos,
- configuraciones colapsadas,
- proximidad extrema entre joints.

---

### 4. Movimiento biomecánico

Buscar:
- movimiento suave,
- asimetría natural,
- regiones de impulso similares a ala.

---

## Objetivo final

Seleccionar una configuración que:

- mantenga continuidad cinemática,
- conserve comportamiento biomimético,
- y sea fácil de implementar posteriormente en CAD.
